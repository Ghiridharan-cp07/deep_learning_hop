##Step 2: Data Collection

###Import Necessary Libraries

```python
from tensorflow import keras
import numpy as np
from tensorflow.keras.datasets import boston_housing
from keras.models import Sequential,load_model
from keras.layers import Dense
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
```

###Load Dataset

```python
(X_train,y_train),(X_test,y_test) = boston_housing.load_data()
```

###Step 3: Data Understanding

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

##Step 4: Data Preparation

```python
std_scaler = StandardScaler()
X_train = std_scaler.fit_transform(X_train)
X_test = std_scaler.transform(X_test)
```

##Step 5: Model Building

```python
model = Sequential([
keras.layers.Input(shape= (13,)),
Dense(64,activation='relu'),
Dense(32,activation='relu'),
Dense(16,activation='relu'),
Dense(1)
])
```

```python
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

##Step 6: Model Training || Model Testing || Model Evaluation

```python
model.fit(x= X_train,y= y_train,batch_size=32,epochs=84,validation_split=0.2,verbose=1)
```

```python
model.summary()
```

```python
model.save('house_price_model.keras')
```

###Model Prediction With My Data

```python
house_model_brain = load_model('house_price_model.keras')
```
