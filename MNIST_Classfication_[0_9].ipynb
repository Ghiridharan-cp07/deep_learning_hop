##Step 2: Data Collection

###Import Necessary Libraries

```python
import keras
from keras.models import Sequential,load_model
from keras.layers import Flatten,Dense
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

```

###Load Dataset

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

##Step 4: Data Preparation

###Let's normalize the Image inputs

```python
X_train.min(),X_train.max()
X_train = X_train/255.0
X_test = X_test/255.0
```

```python
X_train.min(),X_train.max()
X_test.min(),X_test.max()
```

##Step 5: Model Building

###Let's Build Artificial Deep Neural Network(ANN)

```python
# model = Sequential([
#     keras.layers.Input(shape=(28, 28)),
#     Flatten(),
#     Dense(100, activation="relu"),
#     Dense(10, activation="softmax")
# ])
# model.compile(optimizer='rmsprop',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model = Sequential()
model.add(layer = Flatten())
model.add(Dense(units=100,activation='relu'))
model.add(Dense(units=10,activation='softmax'))
model.compile(optimizer='rmsprop',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

##Step 6: Model Training || Step 7: Model Testing || Step 8: Model Evaluation

```python
model.fit(x=X_train,y=y_train,batch_size=32,epochs=10,validation_data=(X_test,y_test))
```

```python
model.summary()
```

###Model Prediction

```python
y_pred = model.predict(X_test)
```
